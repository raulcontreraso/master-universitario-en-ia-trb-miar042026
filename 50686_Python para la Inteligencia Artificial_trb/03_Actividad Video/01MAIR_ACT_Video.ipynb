```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## 01
- Carga de datos de csv (valencia_pollution_dataset.csv) en un DataFrame.
- Mostrar 6 fillas aleatorias
- Mostrar la cantidad de valores nulos por columna

```python
# 1. Cargar el dataset en un DataFrame
df = pd.read_csv('valencia_pollution_dataset.csv')

# 2. Mostrar 6 filas aleatorias
print("--- 6 Filas Aleatorias del Dataset ---")
display(df.sample(6))

# 3. Mostrar la cantidad de valores nulos por columna
print("\n--- Cantidad de Valores Nulos por Columna ---")
print(df.isnull().sum())# Your solution here...
```

## 02
- ¿Cuantas estaciones diferentes hay?
- ¿Cuantas estaciones cuentan con un sensor de monóxido de carbono (co)?
- ¿Cuantas estaciones cuentan con un sensor ozono (o3)?

```python
# 1. ¿Cuántas estaciones diferentes hay?
estaciones_unicas = df['nombre'].nunique()
print(f"Cantidad de estaciones diferentes: {estaciones_unicas}")

# 2. ¿Cuántas estaciones cuentan con un sensor de monóxido de carbono (co)?
# Contamos cuántas estaciones tienen al menos una medición que NO sea nula en 'co'
estaciones_co = df[df['co'].notnull()]['nombre'].nunique()
print(f"Estaciones con sensor de Monóxido de Carbono (co): {estaciones_co}")

# 3. ¿Cuántas estaciones cuentan con un sensor de ozono (o3)?
estaciones_o3 = df[df['o3'].notnull()]['nombre'].nunique()
print(f"Estaciones con sensor de Ozono (o3): {estaciones_o3}")
```

## 03
 - Para cada sensor (co, o3, no2, so2, pm10 y pm25) indicar la estación con la media de medición más alta.

```python
# Lista de sensores a evaluar
sensores = ['co', 'o3', 'no2', 'so2', 'pm10', 'pm25']

print("--- Estación con la media de medición más alta por sensor ---")
for sensor in sensores:
    # Agrupamos por estación, calculamos la media del sensor y obtenemos el índice del valor máximo (idxmax)
    estacion_max = df.groupby('nombre')[sensor].mean().idxmax()
    media_max = df.groupby('nombre')[sensor].mean().max()
    print(f"Sensor {sensor.upper()}: {estacion_max} (Media: {media_max:.2f})")
```

## 04
Mostrar en dos gráficos de barras la media de las mediciones de dióxido de nitrógeno (no2) de cada estación agrupando por el tipo de zona (tipozona). Cada gráfico debe mostrar un tipo de emisión distinto (tipoemision).

```python
# Agrupamos los datos para obtener las medias según tipozona y tipoemisio
df_grouped = df.groupby(['tipozona', 'tipoemisio', 'nombre'], as_index=False)['no2'].mean()

# Creamos una figura con dos subgráficos (uno para cada tipo de emisión)
fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=True)

# Lista de los tipos de emisión únicos para iterar
tipos_emision = df_grouped['tipoemisio'].dropna().unique()

for i, emision in enumerate(tipos_emision):
    # Filtramos los datos para el tipo de emisión actual
    df_sub = df_grouped[df_grouped['tipoemisio'] == emision]
    
    # Creamos el gráfico de barras usando seaborn para agrupar por tipozona de forma automática
    sns.barplot(data=df_sub, x='nombre', y='no2', hue='tipozona', ax=axes[i], palette='Set2', edgecolor='black')
    
    # Configuraciones visuales de cada gráfico
    axes[i].set_title(f"Emisión: {emision}", fontsize=13, fontweight='bold')
    axes[i].set_xlabel("Estaciones de Monitoreo", fontsize=11)
    axes[i].set_ylabel("Media de NO2 (µg/m³)" if i == 0 else "", fontsize=11)
    axes[i].tick_params(axis='x', rotation=45)
    axes[i].grid(axis='y', linestyle='--', alpha=0.7)

plt.suptitle("Media de NO2 por Estación agrupado por Tipo de Zona y Emisión", fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

## 05
Eres un analista de datos y quieres evaluar visualmente si existe una relación entre las mediciones de los sensores (co, o3, no2, so2, pm10 y pm25) y la calidad del aire (calidad_ambiental). ¿Qué visualización propones y porqué?

```python
# Calculamos la matriz de características promedio agrupada por la calidad ambiental
columnas_sensores = ['co', 'o3', 'no2', 'so2', 'pm10', 'pm25']

# Creamos diagramas de caja (Boxplots) para evaluar la distribución de cada sensor según la calidad del aire
plt.figure(figsize=(16, 10))
for i, sensor in enumerate(columnas_sensores, 1):
    plt.subplot(2, 3, i)
    sns.boxplot(data=df, x='calidad_am', y=sensor, palette='Spectral')
    plt.title(f"Distribución de {sensor.upper()} vs Calidad Aire")
    plt.xlabel("Calidad Ambiental")
    plt.ylabel("Concentración")
    plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```
