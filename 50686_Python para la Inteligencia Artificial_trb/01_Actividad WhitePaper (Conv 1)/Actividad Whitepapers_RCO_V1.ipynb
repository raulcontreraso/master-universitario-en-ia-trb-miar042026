#### Pipeline 1 - Conjunto de datos sin procesar (raw datasets) → DataFrame inicial

El artículo describe una fase de preparación de datos sin procesar, luego se realiza el preprocesamiento. El siguiente código muestra una posible implementación de ese proceso en Python.

Se crean Datos sinteticos referentes a marcadores de salud dentro de rangos reconocidos para cada marcador y se crea un dataframe para poder visualizar de manera tabular los datos.

En la celda del  notebook se muestra una posible implementación de este flujo de trabajo utilizando Python. En primer lugar, se cargan los datos clínicos en una estructura manipulable, representando el conjunto de datos en bruto.  

En primer lugar, se cargan los datos clínicos en una estructura manipulable, representando el conjunto de datos en bruto.

```python
#----------------------------------------------------------------------------------
#
# Pipeline 1 - Conjunto de datos sin procesar(raw datasets) → DataFrame inicial
#
#-----------------------------------------------------------------------------------  


# Línea 1— Importar todas las librerias necesarias
#
import pandas as pd  # Tablas 
import numpy as np   # Números y generación de números aleatorios 


# Línea 2 — Reproducibilidad (Reproducibility) ejecuta el mismo código bajo las mismas condiciones 
#           Se obtienen los mismos resultados. Reproducibilidad del experimento, no del resultado clínico.
np.random.seed(42) #Genera los mismos numero aleatorios.

# Linea 3 — Cuantos Pacientes se generaran de forma sintetica
n_patients = 20    #  El alcance de la muestra

# Línea 4 — Diccionario de Datos (Data dictionary)
data = { 
#      a) — Genera números aleatorios de marcadores medicos 
#      uniform(a, b) → Genera numeros aleatorios dentro de un rango
     
   "patient_id": np.arange(1, n_patients + 1), 
   "patient_name": [f"Patient_{i}" for i in range(1, n_patients + 1)], 
      
   "blood_sugar_mg_dl": np.random.uniform(70, 200, n_patients),  # Los rangos pretenden simular los valores normales de un marcador de Salud
   "max_heart_rate_bpm": np.random.uniform(90, 190, n_patients), 
     
   "total_cholesterol_mg_dl": np.random.uniform(120, 300, n_patients), 
   "ldl_cholesterol_mg_dl": np.random.uniform(60, 220, n_patients), 
   "hdl_cholesterol_mg_dl": np.random.uniform(30, 100, n_patients), 
   "triglycerides_mg_dl": np.random.uniform(50, 400, n_patients), 
   "lipoprotein_a_mg_dl": np.random.uniform(5, 150, n_patients), 

   #b) —  Columnas binarias, representan estados SI/NO (Binary columns 
   #choice) “For each patient, randomly pick either 'yes' or 'no'.”
   "diabetes": np.random.choice(["yes", "no"], n_patients), 
   "smoker": np.random.choice(["yes", "no"], n_patients), 
   } 

      
#Línea 5 — Crea la matriz tabular numerica:El modelo ML requiere de una matriz numérica en forma tabular (filas × columnas). 
#          El DataFrame transforma el diccionario a una forma tabular. 
df = pd.DataFrame(data)  

#Línea 6 —   filtra el DataFrame y se queda solo con las columnas numéricas (int, float: edad, colesterol, presión, etc.).
#            .columns: extrae los nombres de esas columnas, no los valores. numeric_cols es una lista de nombres de columnas numéricas.
numeric_cols = df.select_dtypes(include="number").columns  
 

#Línea 7 — Redondeo(Rounding)
#          Opcional: para mejor visualización(readability), menos decimales, más fácil de leer
df[numeric_cols] = df[numeric_cols].round(2) 


#Línea 8 — Grabar el Archivo generado de datos sinteticos
#          Save to Excel 
df.to_excel("synthetic_clinical_data.xlsx", index=False) 


#Línea 9  — Despliega la tabla           
df.head() # Muestra las 5 filas, la estructura, y los tipos de datos

```


#### Pipeline 2 - Procesamiento de Datos (data preprocessing) → Limpieza + Selección 
A continuación, se realiza una fase de preprocesamiento que permite tratar los datos y preparar las variables para su análisis. En este punto, los datos ya están en una estructura tabular consistente. 

En este paso se comienzan a transformar los datos con el fin de prepararlos para el aprendizaje automático. 

En este caso las variables que representan estados se transforman a numericas para su futuro procesamiento, para los pacientes con o sin diabetes y los fumadores se representarán con 1 o 0 según sea el caso.



```python
#----------------------------------------------------------------------------------
#
# Pipeline 2 - Procesamiento de Datos (data preprocessing) → Limpieza + Selección 
#               Codificación
#-----------------------------------------------------------------------------------  

#
# Codificación (Encoding) Transforma variables no numéricas (por ejemplo, sí/no, categorías) y las convierte en valores numéricos
#                        para que puedan ser utilizadas por modelos de aprendizaje automático. 
 

#Línea 1 — Codificacion para valores Binarios "yes" = 1, "no" =  0
print("ANTES del encoding:") #Las columnas categóricas se visualizan antes y después del proceso de codificación para verificar 
                             #la correcta transformación de valores textuales en representaciones numéricas.

#df[["diabetes", "smoker"]].head()
display(df[["diabetes", "smoker"]].head()) # display() es propio de Jupyter y se usa solo para visualización en notebook.


# Codificación(Encoding) 
df["diabetes"] = df["diabetes"].map({"yes": 1, "no": 0}) # Ocurre la codificación (encoding)..map({...}) reemplaza valores "yes" → 1 "no" → 0
df["smoker"] = df["smoker"].map({"yes": 1, "no": 0})

print("DESPUÉS del encoding:")
#df[["diabetes", "smoker"]].head()
display(df[["diabetes", "smoker"]].head())

```

#### Pipeline 3 - Normalización → Scaling (StandardScaler) 
 
Luego del preprocesamiento de datos se continua  con técnicas de transformación (escalado/normalización), coherentes con la etapa de descrita en el artículo.

Tras el escalado, los valores representan la distancia de cada observación respecto a la media en unidades de desviación estándar, por lo que pueden tomar valores negativos o superiores a uno sin que ello suponga un error.

    - Si el valor escalado es −1.0 → está 1 desviación estándar por debajo de la media.
    - Si el valor escalado es −1.5 → está 1.5 desviaciones estándar por debajo de la media.

Los valores escalados se interpretan en relación con la media y la desviación estándar del conjunto de datos analizado, por lo que representan posiciones relativas dentro de la muestra.

El escalado se realiza por columna, de forma independiente para cada variable, garantizando que las comparaciones se efectúen entre características homogéneas y no entre magnitudes distintas dentro de una misma observación.

El comando fit_transform() es el encargado del escalado. Este escalado es por columna, no por fila.
 

```python
#----------------------------------------------------------------------------------
#
# Pipeline 3 - Normalización  → Scaling (StandardScaler) 
#
#----------------------------------------------------------------------------------- 

#
#  Escalado (Scaling) 
#    El escalado es una técnica de preprocesamiento que transforma variables numéricas para que estén en la misma escala,
#       evitando que unas dominen a otras solo por su magnitud.
 

# Línea 1 — Importa las librerias necesarias para la normalizacion
from sklearn.preprocessing import StandardScaler

# Línea 2 — Identificar columnas numéricas — Identify which columns should be scaled
#      Solo las variables numéricas se pueden escalar
#      Variables categóricas o texto no se  pasan por el scaler

numeric_df = df.select_dtypes(include="number") #todas las columnas numéricas del DataFrame.


# Línea 3 — Excluye variables de estado (binary states) / 
continuous_df = numeric_df.drop(columns=["diabetes", "smoker"]) #Quita variables binarias de estado {(0/1),(presencia/ausencia)} del escalado 
                                                               
                                                 
                                                                
# Línea 4 — Aplicar El Escalado  StandardScaler
#      Es una técnica de normalización que centra los datos en 0, ajusta la desviación estándar a 1
#      Escala cada columna de forma independiente, toma una columna a la vez, calcula la media y la desviación estándar 
#      Estos calculos los realiza a una columna, la transforma y pasa a la siguiente.
#      Ejemplo: colesterol → se estandariza usando media y desviación del colesterol y devuelve un array NumPy
#      Centra los datos en 0, ajusta la desviación estándar a 1 
#      Ejemplo del Escalado/normalizacion:  Cuántas desviaciones estándar está ese valor respecto a la media:
#           0 → valor promedio,  
#           negativo → por debajo del promedio
#           positivo → por encima del promedio
#      Un valor escalado de:
#          +1 → una desviación estándar por encima de la media
#          +2 → dos desviaciones estándar
#         +3 → muy lejos de la media (posible outlier)
#                     

scaler = StandardScaler() 
scaled_array = scaler.fit_transform(continuous_df) # escalado se hace por columna, no por fila # Devuelve un array
#print(scaled_array) 


# Línea 5 — Volver a forma tabular /  Put scaled data back into a DataFrame
#           Lo convertimos en una tabla para que sea legible  
scaled_df = pd.DataFrame(
scaled_array,columns=continuous_df.columns,  index=df.index)

  
# Línea 6 — Inspección del resultado  Inspect the result  
           #Tras el proceso de escalado, los valores no se redondean para preservar la precisión necesaria 
           #utilizámos el redondeo únicamente con fines de visualización.
#scaled_df.head() 
scaled_df.round(2).head() # Se muestran redondeados para facilitar la lectura pero internamente mantienen su precisión

```
