**a) Método de Laplace (Expansión por Cofactores)**
   
   Laplace se basa en reducir el problema: $\det(A) = \sum_{j=0}^{n-1} (-1)^{0+j} \cdot a_{0j} \cdot \det(M_{0j})$
   
  -   $a_{0j}$: Elemento en la fila 0, columna $j$.
  
  -   $(-1)^{0+j}$: El "signo de tablero de ajedrez" (alterna entre + y -).
  
  -   $M_{0j}$: La submatriz resultante de eliminar la fila 0 y la columna $j$.

```
#-------------------------------------------------------------------------------
# Metodo de Lapace
#-------------------------------------------------------------------------------

def determinante_recursivo_laplace(matriz, nivel=0, v_print=0):
    """
    Calcula el determinante de forma recursiva (Laplace)
    """
    # Identación visual para el reporte jerárquico
    indent = "  " * nivel
    n = matriz.shape[0]
    # Propiedad de los arreglos de NumPy,devuelve una tupla que indica las dimensiones de la matriz.
    # Por ejemplo,  una matriz de 3 x 3 , shape devuelve (3, 3).[0]:
    # AEl 0, le indica a Python que tome el primer valor de esa tupla. Es decir, nos da el número de filas de la matriz.
    if v_print == 1:
        print(f"\n{indent}=== Tamaño de la matriz (Matriz {n}x{n}) ===")


    # 1. CASO BASE: Matriz 2x2 se detiene la llamada recursiva
    if n == 2:
        return matriz[0, 0] * matriz[1, 1] - matriz[0, 1] * matriz[1, 0]
          # Sub_Matriz =  [3, 4]
          #               [6, 2]
          # = 3 x 2 - 4 x 6
    det = 0
    if v_print == 1:
        print(f"\n{indent}  == Iniciando expansión nivel {nivel} (Matriz {n}x{n}) ===")

    # 2. CASO RECURSIVO: Expandir por la primera fila
    for j in range(n):
        cofactor = matriz[0, j]
           # Término 1: Cofactor B[0,0] = 0
           # Matriz =   [5, 2, 9]
           #            [1, 3, 4]
           #            [8, 6, 2]
           # los cofactores segun valor de J son 5, 2, 9

        # Slicing dinámico: Elimina fila 0 y columna j
        submatriz = np.delete(np.delete(matriz, 0, axis=0), j, axis=1)
        # Paso 1 (Lo de adentro): np.delete(matriz, 0, axis=0):Toma la matriz,
        # busca el índice 0 (la primera fila),
        # y la elimína (axis=0 significa filas)".
        # Paso 2 (Lo de afuera): np.delete(..., j, axis=1), Toma lo que resultó del paso anterior y,
        # a eso, quítale la columna j (axis=1 significa columnas).
        # axis=0: Es el eje vertical. Cuando  se usa, afecta las filas (cortando horizontalmente).
        # axis=1: Es el eje horizontal. Cuando se  usa, afectas las columnas (cortando verticalmente).
        # Delete crea un nuevo arreglo eliminando las filas y columnas indicadas.
        # Sub_Matriz =  [3, 4]
        #               [6, 2]
        # Ejemplo
        # temporal = np.delete(matriz, 0, axis=0)    # Elimina fila
        # submatriz = np.delete(temporal, j, axis=1) # Elimina columna

        # LLAMADA RECURSIVA: La función se llama a sí misma
        det_sub = determinante_recursivo_laplace(submatriz, nivel + 1)

        # Aplicamos la fórmula: (-1)^j * cofactor * Det(submatriz)
        term = ((-1) ** j) * cofactor * det_sub
        det += term

        # 3. PASO A PASO: Impresión de Sub_Matrices
        if v_print == 1:
           print(f"{indent}Término {j+1} (Cofactor {cofactor:.0f}):")
           print(f"{indent}  Submatriz:")

        # BUCLE DINÁMICO: Esto evita el error IndexError
        for fila in submatriz:
            if v_print == 1:
               print(f"{indent}  {fila}")


        signo = "+" if j % 2 == 0 else "-" # Parte de la formula matematica que controla el signo
        # ejemplo En una matriz de 3x3, los índices de las columnas son 0, 1 y 2:
        # Columna j=0: Par   (0) (+)
        # Columna j=1: Impar (1) (-)
        # Columna j=2: Par   (2) (+)
        # Asigna a la variable signo el valor de '+' si el resto de dividir j entre 2 es igual a 0,
        # de lo contrario, asígnale '-' }
        # j % 2 == 0? Es la forma matemática de saber si un número es par.
        #En la expansión de Laplace, los términos alternan de signo según la posición.
        #Esto hace que si estás en la columna 0, 2, 4 (pares), el signo sea positivo,
        # y si estás en la columna 1, 3, 5 (impares), el signo sea negativo.
        #Este el mecanismo que implementa la regla de los signos de los cofactores

        if v_print == 1:
           print(f"{indent}  Operación: {signo} {cofactor:.0f} * ({det_sub:.2f})")
           print(f"{indent}  Resultado Término {j+1} = {term:.4f}")

    return det


```

Respuesta:
- Intercambiar una fila (o columna) por otra fila (o columna).
El determinante representa el volumen orientado o área, en 2D, definido por los vectores fila de la matriz. Al intercambiar dos filas, se invierte el sentido de ese volumen en el espacio por lo que el signo de su medida (el determinante) se invierte.

Si $A$ es la matriz original y $B$ es la matriz resultante tras intercambiar dos filas, la relación es:

$$\det(B) = -1 \cdot \det(A)$$

Ejemplo:

Sea $A = \begin{bmatrix} 1 & 2 \\ 3 & 4 \end{bmatrix}$.

Su determinante es: $(1 \cdot 4) - (2 \cdot 3) = 4 - 6 = \mathbf{-2}$.

al  intercambiar las filas, obtenemos $B = \begin{bmatrix} 3 & 4 \\ 1 & 2 \end{bmatrix}$.Su determinante es: $(3 \cdot 2) - (4 \cdot 1) = 6 - 4 = \mathbf{2}$.

Resultado: $2 = -1 \cdot (-2)$. El valor cambió de $-2$ a $2$.

- Sumar a una fila (o columna) otra fila (o columna) multiplicada por un escalar α.

 Teóricamente, esta operación preserva el volumen orientado (el determinante) de la matriz. Al añadir un múltiplo de una fila a otra, se está aplicando una "transformación"  que desplaza los vectores fila sin cambiar el área o volumen que encierran, por lo que el determinante permanece inalterado.

Si $A$ es la matriz original y $B$ es la matriz resultante tras realizar la operación $R_j \leftarrow R_j + \alpha R_i$, entonces:

$$\det(B) = \det(A)$$

**d) Investiga sobre el metodo de eliminacion de Gauss con pivoteo parcial e

**Método de Gauss (Eliminación)**
La filosofía de Gauss: "Triangular la Matriz"
Busca que todos los números que están debajo de la diagonal se vuelvan cero:

$$\begin{bmatrix}
Pivote & * & * \\
0 & Pivote & * \\
0 & 0 & Pivote
\end{bmatrix}$$(Donde * es cualquier número y 0 son las zonas que "Eliminadas").

$$A = \begin{bmatrix} 5 & 2 & 9 \\ 1 & 3 & 4 \\ 8 & 6 & 2 \end{bmatrix}$$



Paso 1: Columna 1
-    Pivote Inicial ($A_{11}$): El elemento en la posición diagonal es 5.

-   ¿Es el más grande de su columna? Buscamos en la columna 1: $|5|$, $|1|$, $|8|$. El valor absoluto más grande es 8 (fila 3).


-   Pivoteo Parcial (Intercambio): Intercambiamos la Fila 1 ($R_1$) con la Fila 3 ($R_3$) para garantizar la estabilidad numérica. Contador de intercambios: $1$ (por lo tanto, al final multiplicaremos el resultado por $-1$).
La matriz queda así:

$$A' = \begin{bmatrix} \mathbf{8} & 6 & 2 \\ 1 & 3 & 4 \\ 5 & 2 & 9 \end{bmatrix}$$


-   Eliminación debajo del nuevo pivote (8):  Fila 2 ($R_2$): El factor de eliminación es $1/8 = 0.125$.

$$R_2 \leftarrow R_2 - 0.125 \cdot R_1$$


$$R_2 = [1 - 0.125(8), \quad 3 - 0.125(6), \quad 4 - 0.125(2)] = [0, \quad 2.25, \quad 3.75]$$


Fila 3 ($R_3$): El factor de eliminación es $5/8 = 0.625$.$$R_3 \leftarrow R_3 - 0.625 \cdot R_1$$

$$R_3 = [5 - 0.625(8), \quad 2 - 0.625(6), \quad 9 - 0.625(2)] = [0, \quad -1.75, \quad 7.75]$$

Caluladora:

- 5 - 0.625 * 8 = 0
- 2 - 0.625 * 6 = -1.75
- 9 - 0.625 * 2 = 7.75



La matriz tras esta fase es:$$\begin{bmatrix} 8 & 6 & 2 \\ 0 & \mathbf{2.25} & 3.75 \\ 0 & -1.75 & 7.75 \end{bmatrix}$$


Paso 2: Columna 2

-   Pivote Actual ($A_{22}$): El elemento en la diagonal es 2.25.
-   ¿Es el más grande de su columna hacia abajo? Comparamos $|2.25|$ con $|-1.75|$. Como $2.25$ es el mayor, no hace falta realizar un intercambio.
-   Eliminación debajo de 2.25:
 Fila 3 ($R_3$): El factor es $-1.75 / 2.25 \approx -0.7778$.

 $$R_3 \leftarrow R_3 - (-0.7778) \cdot R_2 \implies R_3 \leftarrow R_3 + 0.7778 \cdot R_2$$

 $$R_3 = [0, \quad -1.75 + 1.75, \quad 7.75 + 0.7778(3.75)] = [0, \quad 0, \quad 10.6667]$$

 La matriz final en su forma Triangular Superior ($U$) es:$$U = \begin{bmatrix} 8 & 6 & 2 \\ 0 & 2.25 & 3.75 \\ 0 & 0 & 10.6667 \end{bmatrix}$$



Cálculo del Determinante

Multiplicamos los elementos de la diagonal principal y aplicamos el cambio de signo por el intercambio de filas realizado en el Paso 1:

$$\det(A) = (-1)^{\text{intercambios}} \cdot (u_{11} \cdot u_{22} \cdot u_{33})$$

$$\det(A) = (-1)^1 \cdot (8 \cdot 2.25 \cdot 10.6667)$$

$$\det(A) = -1 \cdot (18 \cdot 10.6667) = -1 \cdot 192 = -192$$

```
#--------------------------------------------------------------------------------
# Metodo de eliminacion de Gauss
#  con pivoteo parcial para escalonar una matriz,
#  es decir, convertirla en una matriz triangular inferior)
#--------------------------------------------------------------------------------

# 1. Función para escalonar (Pivoteo parcial + Eliminación)
def escalonar_matriz_pivoteo(matriz):
    A = matriz.astype(float)
    n = A.shape[0]
    intercambios = 0

    for i in range(n):
        # Pivoteo parcial
        max_index = i + np.argmax(np.abs(A[i:, i]))
        if i != max_index:
            A[[i, max_index]] = A[[max_index, i]]
            intercambios += 1

        # Eliminación
        for j in range(i + 1, n):
            if A[i, i] == 0: continue
            factor = A[j, i] / A[i, i]
            A[j, i:] -= factor * A[i, i:]

    return A, intercambios
```

```
# Ejemplo de uso:
matriz_2 = np.array([
    [8, 6, 2],
    [1, 3, 4],
    [5, 2, 9]
], dtype=float)


print("\nMatriz  x :             \n", matriz_2)
matriz_escalonada = escalonar_matriz_pivoteo(matriz_2)
# nota: Si la función devuelve (matriz, intercambios) (que es una tupla), y se  asigna a una sola variable
# matriz_escalonada ya no será una matriz, sino una tupla que contiene ambos valores: (matriz_real, numero_intercambios).
# por eso en la linea de abajo se usa "_" . las tuplas ni se pueden modificar
print(f"Matriz escalonada:\n{matriz_escalonada}")
```

**e) ¿Como se podria calcular el determinante de una matriz haciendo beneficio
de la estrategia anterior y del efecto de aplicar las operaciones elementales pertinentes? Implementa una nueva funcion, determinante gauss, que calcule el determinante de una matriz utilizando eliminacion gaussiana. [0.5 puntos]**

```
#--------------------------------------------------------------------------------
#
# Determinante de una matriz utilizando eliminacion gaussiana.
#
#--------------------------------------------------------------------------------

# 2. Función para calcular el determinante (usando la anterior)
# Definición RE-ASEGURADA de la función de determinante
def calcular_determinante_gauss(matriz):
    matriz_escalonada, intercambios = escalonar_matriz_pivoteo(matriz)
    det = np.prod(np.diag(matriz_escalonada)) * ((-1) ** intercambios)
    return det
```

### - Definición Recursiva (Expansión por Cofactores / Laplace)
*   **Complejidad:** $O(n!)$ (Factorial).
*   El algoritmo genera una ramificación recursiva de $n$ subproblemas por cada paso, lo que causa un crecimiento explosivo.Es  ineficiente y a partir de $n=10$, el número de operaciones alcanza millones; para matrices mayores, es inviable computacionalmente.

### - Eliminación de Gauss (con Pivoteo Parcial)
*   **Complejidad:** $O(n^3)$ (Cúbica).
*   Se basa en tres ciclos anidados que reducen la matriz a forma triangular. El esfuerzo crece de forma polinómica, no exponencial. Es   Altamente eficiente y capaz de manejar matrices grandes en tiempo real.

Para el cálculo del determinante de matrices $A_n \in \mathbb{R}^{n \times n}$, se analizaron tres enfoques distintos:

1. Método Recursivo (Laplace): Implementa una expansión por cofactores. Su complejidad es $O(n!)$, lo que implica un crecimiento factorial. Es el método más ineficiente. Incluso para matrices pequeñas ($n > 10$), el tiempo de ejecución aumenta drásticamente, haciendo que sea inviable para aplicaciones de ciencia de datos.

2. Método de Eliminación Gaussiana: Reduce la matriz a una forma triangular superior mediante eliminación. Su complejidad es $O(n^3)$. Es mucho más eficiente que la recursión. Para un $n$ moderado, es la mejor opción.

3. Función Preprogramada (numpy.linalg.det): Utiliza algoritmos de álgebra lineal altamente optimizados. Es el método estándar mas utilizzado.

## **Apartado a) Implementación general del Descenso de Gradiente**

El algoritmo del descenso de gradiente es un método iterativo de optimización de primer orden utilizado para encontrar mínimos locales de una función. La regla de actualización para un vector de parámetros $x$ en la iteración $k+1$ viene dada por:

$$x_{k+1} = x_k - \gamma \nabla f(x_k)$$

Donde $\gamma$ es el ratio de aprendizaje (*learning rate*) y $\nabla f(x_k)$ es el gradiente de la función en el punto actual. El algoritmo se detiene cuando la norma euclídea del gradiente es menor que una tolerancia predefinida ($||\nabla f(x)||_2 < \text{tol}$) o cuando se alcanza el número máximo de iteraciones (`maxit`).

```
import numpy as np

def descenso_gradiente(grad_f, x0, gamma, tol=1e-12, maxit=1e5):
    """
    Implementación del algoritmo de descenso de gradiente para n variables.
    """
    # Convertimos el punto inicial a un array de numpy de tipo flotante
    x = np.array(x0, dtype=float)

    for i in range(int(maxit)):
        grad = grad_f(x)
        norma_grad = np.linalg.norm(grad, ord=2)

        # Criterio de parada
        if norma_grad < tol:
            print(f"Convergencia alcanzada en la iteración {i}. Norma del gradiente: {norma_grad:.2e}")
            return x

        # Actualización
        x = x - gamma * grad

    print(f"Se alcanzó el número máximo de iteraciones ({maxit}) sin cumplir la tolerancia.")
    return x
```

## **Apartado b) Estudio de la función de una variable** $f(x) = 3x^4 + 4x^3 - 12x^2 + 7$

### **Estudio Analítico Preliminar**
1. **Primera derivada (Gradiente):**
   $$f'(x) = 12x^3 + 12x^2 - 24x = 12x(x^2 + x - 2) = 12x(x + 2)(x - 1)$$
   Puntos críticos en $x = 0$, $x = 1$ y $x = -2$.

2. **Segunda derivada (Clasificación):**
   $$f''(x) = 36x^2 + 24x - 24$$
   * $x = 0 \implies f''(0) = -24 < 0 \implies$ **Máximo local** en $(0, 7)$.
   * $x = 1 \implies f''(1) = 36 > 0 \implies$ **Mínimo local** en $(1, 2)$.
   * $x = -2 \implies f''(-2) = 72 > 0 \implies$ **Mínimo global** en $(-2, -25)$.

```
# Definimos el gradiente de f(x)
def grad_f(x):
    return np.array([12*x[0]**3 + 12*x[0]**2 - 24*x[0]])

tol_b = 1e-12
maxit_b = 1e5
x0_b = [3.0]

print("--- EXPERIMENTO I (gamma = 0.001) ---")
sol_I = descenso_gradiente(grad_f, x0=x0_b, gamma=0.001, tol=tol_b, maxit=maxit_b)
print(f"Resultado: x = {sol_I[0]:.6f}\n")

print("--- EXPERIMENTO II (gamma = 0.01) ---")
sol_II = descenso_gradiente(grad_f, x0=x0_b, gamma=0.01, tol=tol_b, maxit=maxit_b)
print(f"Resultado: x = {sol_II[0]:.6f}\n")

print("--- EXPERIMENTO IV (gamma = 0.1) ---")
try:
    sol_IV = descenso_gradiente(grad_f, x0=x0_b, gamma=0.1, tol=tol_b, maxit=maxit_b)
    print(f"Resultado: x = {sol_IV[0]}\n")
except Exception as e:
    print(f"Error detectado por divergencia: {e}\n")

print("--- EXPERIMENTO V (x0 = 0) ---")
sol_V = descenso_gradiente(grad_f, x0=[0.0], gamma=0.001, tol=tol_b, maxit=maxit_b)
print(f"Resultado: x = {sol_V[0]:.6f}\n")
```

### **Interpretación de los Resultados de la Función de Una Variable**

* **III) Contraste Experimentos I y II:**
  Con $\gamma = 0.001$, los pasos son cortos y el algoritmo cae de forma segura en el mínimo local más cercano a su punto de partida ($x \approx 1.0$). Al aumentar el ratio a $\gamma = 0.01$, el primer paso es tan grande que "salta" el bache del mínimo local y del máximo, aterrizando directamente en la cuenca de atracción del mínimo global ($x \approx -2.0$). El *learning rate* altera drásticamente el destino en superficies no convexas.

* **IV) Interpretación del Experimento IV ($\gamma = 0.1$):**
  Un ratio excesivamente alto provoca que el tamaño del paso sea mayor que el ancho de la propia cuenca de optimización. El algoritmo empieza a **oscilar de manera divergente**, alejándose cada vez más del origen en cada iteración hasta sufrir un desbordamiento numérico (*Overflow/NaN*).

* **V) Interpretación del Experimento V ($x_0 = 0$):**
  **No es un resultado deseable.** El algoritmo se queda atrapado instantáneamente en un **máximo local**. Esto ocurre porque en cualquier punto crítico el gradiente es cero ($f'(0) = 0$), lo que anula el paso de actualización ($x_{k+1} = x_k - \gamma \cdot 0$). El descenso de gradiente clásico es incapaz de distinguir un máximo de un mínimo si empieza exactamente sobre él.

## **Apartado c) Estudio de la función de varias variables** $g(x,y) = x^2 + y^3 + 3xy + 1$

### **Estudio Analítico Preliminar**
1. **Gradiente ($\nabla g$):**
   $$\nabla g(x,y) = \begin{pmatrix} 2x + 3y \\ 3y^2 + 3x \end{pmatrix} = \mathbf{0}$$
   Puntos críticos en **$P_1 = (0,0)$** y **$P_2 = (-2.25, 1.5)$**.

2. **Matriz Hessiana ($H_g$):**
   $$H_g(x,y) = \begin{pmatrix} 2 & 3 \\ 3 & 6y \end{pmatrix}$$
   * **Para $P_1(0,0)$:** $\det(H_g) = -9 < 0 \implies$ **Punto de Silla (Saddle Point)**.
   * **Para $P_2(-2.25, 1.5)$:** $\det(H_g) = 9 > 0$ y $2 > 0 \implies$ **Mínimo Local**.

```
# Definimos el gradiente de g(x, y)
def grad_g(v):
    x, y = v[0], v[1]
    return np.array([2*x + 3*y, 3*y**2 + 3*x])

gamma_c = 0.01
tol_c = 1e-12
maxit_c = 1e5

print("--- EXPERIMENTO 1 (Punto inicial [-1, 1]) ---")
sol_c_I = descenso_gradiente(grad_g, x0=[-1.0, 1.0], gamma=gamma_c, tol=tol_c, maxit=maxit_c)
print(f"Resultado: P = ({sol_c_I[0]:.4f}, {sol_c_I[1]:.4f})\n")

print("--- EXPERIMENTO 2 (Punto inicial [0, 0]) ---")
sol_c_II = descenso_gradiente(grad_g, x0=[0.0, 0.0], gamma=gamma_c, tol=tol_c, maxit=maxit_c)
print(f"Resultado: P = ({sol_c_II[0]:.4f}, {sol_c_II[1]:.4f})")
```

### **III) Explicación y Contraste de Resultados en $g(x,y)$**

* **Desde $x_0 = (-1,1)$:** El algoritmo converge con éxito al punto $(-2.25, 1.50)$. Esto valida experimentalmente el análisis matemático, ya que este punto demostró ser un **mínimo local** real (estable).

* **Desde $x_0 = (0,0)$:** El algoritmo se congela en la iteración 0. **No es un resultado deseable** porque $(0,0)$ es un **punto de silla**. En este tipo de puntos, la función tiene pendientes de bajada en ciertas direcciones, pero como el gradiente en la coordenada exacta es estrictamente nulo ($\nabla g = 0$), el algoritmo clásico se detiene asumiendo erróneamente que ha terminado con éxito.
